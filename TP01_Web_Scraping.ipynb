```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
```

```python
base_url = "https://www.azquotes.com/author/10823-Friedrich_Nietzsche"
```

```python
url = base_url + "?p=1"

response = requests.get(url)

print("Status code:", response.status_code)
```

```python
print(response.text)
```

```python
soup = BeautifulSoup(response.text, "html.parser")
```

```python
quotes = soup.select("a.title")

print("Quotes on the first page:", len(quotes))

for quote in quotes:
    print(quote.get_text(strip=True))
```

```python
all_quotes = []

for page in range(1, 101):

    url = base_url + f"?p={page}"

    response = requests.get(url)

    soup = BeautifulSoup(response.text, "html.parser")

    quotes = soup.select("a.title")

    for quote in quotes:
        all_quotes.append(quote.get_text(strip=True))
        print("Total quotes:", len(all_quotes))
```

```python
df = pd.DataFrame(all_quotes, columns=["Quote"])
```

```python
df.to_csv("quotes.csv", index=False, encoding="utf-8-sig")
```
